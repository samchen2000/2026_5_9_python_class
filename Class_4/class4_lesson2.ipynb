```python
try:
    with open("note.txt", "r", encoding="utf-8") as file:
        content = file.read()
        print(content)
except Exception as e:
    print("發生錯誤", e)
```

```python
# open("student.txt") 引數值的呼叫,必需依順序
try:
    with open("note2.txt", "r", encoding = "utf-8") as file:
        content_1 = file.read()
        print(content_1)
except Exception as e:
    print("使用錯誤", e)
```

```python
# open(file="student.txt") 引數名稱的呼叫,可以不依順序,只要依照名稱
file_1 = "note2.txt"
try:
    with open(file_1, encoding = "utf-8", mode = "r") as file:
        content_2 = file.read()
        print(content_1)
except Exception as e:
    print("使用錯誤", e)
```

```python
# open(file="student.txt") 混合的呼叫,可以不依順序,只要依照名

try:
	with open("note2.txt",encoding="utf-8",mode="r") as file:
			data = file.read()
			print(data)
except Exception as e:
	print("發生錯誤:", e)
```
