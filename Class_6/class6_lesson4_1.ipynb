```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 4*np.pi, 1000)
y1 = np.sin(x)
y2 = np.sin(2*x)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x, y1, label='sin(x)', linewidth=2)
ax.plot(x, y2, label='sin(2x)', linewidth=2)

ax.set_title('Sine Waves')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True)
ax.legend()

plt.show()
```
