```python
import numpy as np
import pandas as ps
import math
import matplotlib.pyplot as plt
test_1 = np.array([[10, 20, 30], [3, 6, 9]])
test_2 = np.array([2, 4, 6])
test_3 = (test_1 * 2) + test_2
test_4 = np.arange(0, 2.5, 0.1)
test_5 = test_4 * math.pi
test_y1 = np.sin(test_5)
test_y2 = np.cos(test_5)
print(len(test_y1))
fig = plt.figure()
axes = fig.add_subplot(1, 1, 1)
axes.plot(test_4, test_y1)
axes.plot(test_4, test_y2)
plt.show()
print(test_3)
print(test_y1)
print(test_y2)
```

```python
#step1 : load numpy , 並取別名 np
import numpy as np
#step2 : load matplotlib.pyplot , 並取別名 plt
import matplotlib.pyplot as plt

#step3 : 使用 np.arange() 產生0~5 (不含5) 間隔0.1 的陣列存入 t 
t = np.arange(0, 5, 0.1)

y1 = np.sin(t * np.pi)
y2 = np.cos(t * np.pi)

figure = plt.figure()
ax1 = figure.add_subplot(1, 2, 1)
ax1.plot(t, y1, linestyle = "--", color = "green")
ax1 = figure.add_subplot(1, 2, 2)
ax1.plot(t, y2, linestyle = "--", color = "red")
plt.show
```
