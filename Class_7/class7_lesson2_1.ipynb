```python
import matplotlib.pyplot as plt

labels = ['Nokia', 'Apple', 'Sansung', 'Lumia']
values = [20, 10, 30, 12]
colors = ['yellow', 'green', 'red', 'blue']

figure = plt.figure()
axes = figure.add_subplot()
axes.pie(values, 
         labels=labels,
         colors = colors, 
         explode=[0.3,0 ,0 ,0],
         shadow=True, 
         autopct="%2.1f%%", 
         startangle=180)
plt.show()
```
