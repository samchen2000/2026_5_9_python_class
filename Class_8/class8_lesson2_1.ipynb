```python
class persons:
    def __init__(self,name):
        ##建立一個attribute
        self.name = name

    ## 自訂實體的merthod
    def introduce(self):
        print(f"哈囉,我的名字是{self.name}")
```

```python
p2 = persons("sam")
p2.introduce()
```

```python
p3 = persons("jerry")
p3.introduce()
```
