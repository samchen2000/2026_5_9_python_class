```python
import numpy as np
from numpy import ndarray
import pandas as pd
from pandas import DataFrame

#scores : ndarray | None = None
#scores_df : DataFrame | None = None

scores:ndarray = np.random.randint(50, 101, size=(100, 5))
scores_df:DataFrame = pd.DataFrame(scores, 
                        columns = ['國文', '英文', '數學', '地理', '歷史'], 
                        index = range(1, 101) )
names_df : DataFrame = pd.read_csv('student_data.csv')
scores_df[['姓名','年齡']] = names_df[['姓名','年齡']].head(n=100).values
scores_df = scores_df[['姓名', '年齡','國文', '英文', '數學', '地理', '歷史']]
#print(type(scores))
#print(scores.shape)
#選取欄
scores_df[['姓名', '年齡']]
#選取列
scores_df.iloc[5]
scores_df.iloc[5:10]
```

```python
scores_df["總分"] = scores_df[['國文', '英文', '數學', '地理', '歷史']].sum(axis=1)
scores_df["平均"] = scores_df[['國文', '英文', '數學', '地理', '歷史']].mean(axis=1)
scores_df["排名"] = scores_df["總分"].rank(ascending=False, method='min')
scores_df.sort_values(by = '排名')
```
